## Suitable Data

### 1. Rental price prediction for apartments in the canton of Zurich

#### Apartment listing data

A suitable primary source is [ImmoScout24](https://www.immoscout24.ch). Each listing can provide:

- Rental price, including additional costs
- Municipality and address or approximate location
- Apartment size in square metres
- Number of rooms
- Number of bathrooms
- Floor and building year
- Availability date
- Apartment type
- Balcony, terrace, garden, elevator, parking, and storage
- Furnished status
- Condition and renovation status
- Heating type and energy information
- Listing date and time on the market

The target variable would be the monthly rental price. Price per square metre could also be calculated for comparison between apartments of different sizes.

#### Municipality-level data

The apartment data should be supplemented with data from the Swiss Federal Statistical Office, such as the *Regionalporträts 2021: Kennzahlen aller Gemeinden*. Relevant variables include:

- Population and population density
- Average age and household size
- Income and employment indicators
- Number of residential buildings and apartments
- Vacancy rate
- Tax level
- Share of foreign residents
- Accessibility and transport information

Additional location data could include distance to train stations, schools, shopping areas, and city centres.

#### Data preparation and combination

Listings should be linked to municipality-level data using the municipality name or official municipality identification number. Address data can also be converted into geographical coordinates to calculate distances and accessibility measures. Duplicate and expired listings should be removed, and categorical variables should be standardized.

### 2. Demand and supply planning for supermarkets in Swiss municipalities

#### Supermarket and shop data

OpenStreetMap is a suitable source for identifying supermarkets and other food retailers. The [OpenStreetMap shop documentation](https://wiki.openstreetmap.org/wiki/DE:Key:shop) describes available tags. Relevant variables include:

- Shop type, such as `supermarket`, `convenience`, or `bakery`
- Name and brand
- Geographic coordinates
- Address and municipality
- Opening hours
- Size or floor area, where available
- Website and contact information
- Accessibility and parking facilities
- Product categories

This data describes the available supply and the geographical distribution of supermarkets. OpenStreetMap data should be supplemented with commercial or official business-register data where possible, because some shops may be missing or have incomplete attributes.

#### Demand data

To estimate demand, additional data is required because OpenStreetMap does not contain sales information. Suitable variables include:

- Historical sales by product, store, and day
- Product prices and promotions
- Stock levels and stockouts
- Inventory deliveries and replenishment quantities
- Product category, brand, and shelf life
- Seasonal and holiday indicators
- Store opening hours
- Customer counts or transaction counts

Sales and inventory data would normally be obtained from supermarket companies or point-of-sale systems.

#### Municipality-level data

The *Regionalporträts 2021* can provide contextual information, including:

- Population and population density
- Age distribution
- Household size
- Income and employment
- Number of households
- Residential and commuter population
- Tourism or seasonal population
- Existing retail and business activity

Other useful data sources include public transport stops, road networks, parking locations, and geographic boundaries.

#### Data preparation and combination

OpenStreetMap shops and supermarket business data should be geocoded and assigned to municipalities. Sales and inventory data can then be linked to each store using a store identifier or geographic location. Municipality-level characteristics can be joined using the official municipality identification number.

The resulting dataset could support demand forecasting, store comparison, stock-level planning, and the identification of municipalities that are underserved or have strong potential for additional supermarket supply.
