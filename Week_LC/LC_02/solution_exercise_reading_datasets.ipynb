### Cars_autoscout24.csv

```python
# Try reading the file with utf-8, fall back to latin1 if that fails
try:
    df = pd.read_csv('Cars_autoscout24.csv', sep=';', encoding='utf-8')
    print('Read with utf-8', '\n')
except UnicodeDecodeError:
    df = pd.read_csv('Cars_autoscout24.csv', sep=';', encoding='latin1')
    print('utf-8 failed, read with latin1 instead', '\n')

# Function to extract numerical values from the Price column
# Note: prices below CHF 1'000 have no thousands separator (e.g. "CHF 899.–"),
# so a regex that requires an apostrophe (r"\d+'\d+") silently returns None
# for those rows. Strip apostrophes first, then extract the leading digits.
def extract_numerical_value(price):
    digits = price.replace("'", '')
    match = re.search(r"\d+", digits)
    return float(match.group()) if match else None

# Apply the function to the 'Price' column and create a new column 'Prices_numeric'
df['Prices_numeric'] = df['Price'].apply(extract_numerical_value)

# Display the DataFrame
print(df)
```

### Histogram of car prices

```python
# Plotting the histogram with car prices
plt.figure(figsize=(6,4))
plt.hist(df['Prices_numeric'], bins=20, color='greenyellow', edgecolor='black')
plt.xlabel('Price (CHF)')
plt.ylabel('Frequency')
plt.title('Histogram of car prices')
plt.grid()
plt.show()
```

```python
# Report min, max and mean price actually observed in this run
print('Min price:  CHF', df['Prices_numeric'].min())
print('Max price:  CHF', df['Prices_numeric'].max())
print('Mean price: CHF', round(df['Prices_numeric'].mean(), 2))
```

### Personalized subset analysis (Task 4)

Example below uses `last_digit = 7` (odd → `Fuel_Type == 'Diesel'`). Replace
`last_digit` with the last digit of your own student ID — an even digit maps
to `Benzin`, an odd digit maps to `Diesel`.

```python
# Replace with the last digit of your own student ID
last_digit = 7

fuel_type = 'Benzin' if last_digit % 2 == 0 else 'Diesel'
print(f'Last digit: {last_digit} --> Fuel_Type: {fuel_type}')

# Filter to the assigned fuel type and recompute Prices_numeric for the subset
df_subset = df[df['Fuel_Type'] == fuel_type].copy()

print(f'Number of cars in subset: {len(df_subset)}')
print(f'Mean price for {fuel_type}: CHF', round(df_subset['Prices_numeric'].mean(), 2))

# Histogram for the subset
plt.figure(figsize=(6,4))
plt.hist(df_subset['Prices_numeric'], bins=20, color='skyblue', edgecolor='black')
plt.xlabel('Price (CHF)')
plt.ylabel('Frequency')
plt.title(f'Histogram of car prices ({fuel_type} only)')
plt.grid()
plt.show()
```
