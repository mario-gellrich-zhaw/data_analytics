```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
simulated_data = rng.normal(loc=50, scale=10, size=1000)

plt.hist(simulated_data, bins=30, edgecolor="black")
plt.title("Histogram of Simulated Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```
