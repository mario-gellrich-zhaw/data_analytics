### Task 5: Why cell order matters

**Observation:** Cells in a notebook share one running kernel, so a cell's output depends on which cells have *actually been run*, and in what order — not on their position in the notebook. If a `print(x)` cell were run *before* the cell defining `x = 1`, it would raise `NameError: name 'x' is not defined`, even though `x = 1` appears above it in the notebook. This is why, before trusting a notebook's outputs (or submitting it), you should use **Kernel → Restart Kernel and Run All Cells**: it guarantees every cell runs top-to-bottom in a fresh kernel, the same way a grader (or your future self) will run it.
