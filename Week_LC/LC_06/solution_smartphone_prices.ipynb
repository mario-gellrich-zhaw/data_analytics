## Libraries and settings

```python
# Libraries
import os
import numpy as np
import pandas as pd
import seaborn as sns
import scipy.stats as stats
import matplotlib.pyplot as plt
from IPython.display import Image

# Ignore warnings
import warnings
warnings.filterwarnings("ignore")

# Show current working directory
print(os.getcwd())
```

## Simulated smartphone data
The data was simulated with `smartphone_data_simulator.ipynb` (Task 1) and saved as `smartphone_data.csv`.

```python
# Read the simulated data from 'smartphone_data.csv' (created with smartphone_data_simulator.ipynb)
df = pd.read_csv('smartphone_data.csv')

# Show the first rows and the number of rows per brand
print(df['brand'].value_counts())
df.head()
```

## Grouped boxplots showing how smartphone prices vary per brand

```python
# Create a grouped boxplot using Seaborn
plt.figure(figsize=(8,2))
plt.title('Smartphone prices by brand')
sns.boxplot(x='price_chf', 
            y='brand', 
            data=df,
            palette='Set2',
            orient='h')
plt.xlabel('Price (CHF)')
plt.ylabel('Brand')
plt.show()
```

## Anlysis of Variance (ANOVA)

```python
# Hypotheses:
# H0: There are no price differences between groups (smartphone brands)
# H1: There are price differences between groups (smartphone brands)

# Significance level alpha
alpha = 0.05

# Create ANOVA
# One group of prices per brand (works for any number of brands)
groups = [g['price_chf'] for _, g in df.groupby('brand')]
fvalue, pvalue = stats.f_oneway(*groups)

# Print results
print(f'F-value: {fvalue:.3f}, p-value: {pvalue:.3g}')

# Interpretation
print('\nInterpretation:')
if pvalue < alpha:
    print(f"The differences in smartphone prices among brands are statistically significant at the {alpha} significance level.")
else:
    print(f"There is no statistically significant difference in smartphone prices among brands at the {alpha} significance level.")
```
